```python
import pandas as pd
import numpy as np

data = {
    "Product_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 102],
    "Product_Name": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones",
                     "Smartphone", "Tablet", "Printer", "Webcam", "Speaker",
                     "Router", "SSD", "Microphone", "Gaming Chair", "Mouse"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics",
                 "Accessories", "Electronics", "Electronics", "Electronics",
                 "Accessories", "Accessories", "Networking", "Storage",
                 "Accessories", "Furniture", "Accessories"],
    "Price": [55000, 800, 1500, 12000, 2500, 75000, 30000, 8500,
              2200, 3500, 4000, 6000, 1800, 15000, 800],
    "Quantity_Sold": [5, 25, 15, 8, 12, 4, 6, 3, 20, 10, 7, 9, 11, 2, 25],
    "Supplier": ["A", "B", "B", "A", "C", "A", None, "C", "B",
                 "C", "A", "B", None, "D", "B"],
    "Rating": [4.5, 4.2, None, 4.7, 4.0, 4.8, 4.3, None, 3.9,
               4.1, 4.4, None, 4.2, 4.6, 4.2]
}

df = pd.DataFrame(data)

print(df)
```

```python
#identify the missing values
print(df.isnull())

missing_count = df.isnull().sum()
print("Missing values in each column:")
print(missing_count)

missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Missing percentage:")
print(missing_percentage.round(2))

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage.round(2)
})

print(missing_summary)
```

```python
#Handling the missing values
df["Supplier"] = df["Supplier"].fillna("Unknown")
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

print(df.isnull().sum())
```

```python
#detect duplicate records
duplicate_rows = df[df.duplicated()]
print("Duplicate records:")
print(duplicate_rows)

print("Total duplicate records:", df.duplicated().sum())

repeated_ids = df["Product_ID"].value_counts()
print("Product ID occurrences:")
print(repeated_ids[repeated_ids > 1])

duplicate_ids = df["Product_ID"].duplicated(keep=False)
print("All records associated with repeated IDs:")
print(df[duplicate_ids])
```

```python
#REmoving duplicate recods
df = df.drop_duplicates()

print("Records after removing exact duplicates:")
print(df)

print("Remaining duplicate records:", df.duplicated().sum())
```

```python
#Group wise data Analysis
df["Revenue"] = df["Price"] * df["Quantity_Sold"]

category_analysis = df.groupby("Category").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity_Sold=("Quantity_Sold", "sum"),
    Average_Price=("Price", "mean"),
    Median_Price=("Price", "median"),
    Product_Count=("Product_ID", "count")
)

category_analysis = category_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
)

print(category_analysis)
```

```python
#highest and lowest 
highest_category = category_analysis["Total_Revenue"].idxmax()
lowest_category = category_analysis["Total_Revenue"].idxmin()

print("Highest revenue category:", highest_category)
print("Highest revenue:", category_analysis.loc[highest_category, "Total_Revenue"])

print("Lowest revenue category:", lowest_category)
print("Lowest revenue:", category_analysis.loc[lowest_category, "Total_Revenue"])
```

```python
#highest priced product
highest_price = df["Price"].max()
print("Highest price:", highest_price)

highest_index = df["Price"].idxmax()
highest_product = df.loc[highest_index]

print("Highest-priced product:")
print(highest_product)

top_five = df.nlargest(5, "Price")

print("Top five highest-priced products:")
print(top_five[["Product_ID", "Product_Name", "Category", "Price"]])
```

```python
print("Dataset information:")
print(df.info())

print("Numerical summary:")
print(df.describe())

print("Missing values after cleaning:")
print(df.isnull().sum())

print("Duplicate rows after cleaning:")
print(df.duplicated().sum())
```
