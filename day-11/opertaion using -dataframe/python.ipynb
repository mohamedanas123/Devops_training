```python
import pandas as pd
import numpy as np

data = {
    "Order_ID": range(1001, 1016),
    "Customer_Name": [
        " Anas ", "Rahul", "Priya ", "Kumar", "Divya",
        "Arun", "Sneha", "Vijay", " Meena", "Ravi",
        "John", "Kavya", "Ajay", "Sara", "Manoj"
    ],
    "City": [
        "Chennai", "chennai", " CHENNAI ", "Madurai", "madurai",
        "Coimbatore", "COIMBATORE", "Chennai", "Madurai", None,
        "Chennai", "Trichy", " trichy ", "Coimbatore", "Chennai"
    ],
    "Product": [
        "Laptop", "Mobile", "Tablet", "Laptop", "Mobile",
        "Tablet", "Laptop", "Mobile", "Tablet", "Laptop",
        "Mobile", "Laptop", "Tablet", "Mobile", "Laptop"
    ],
    "Quantity": [
        2, 1, 3, 2, None, 1, 2, 1, 4, 2, 1, 3, 2, 1, 100
    ],
    "Price": [
        "50000", "15000", "12000", "52000", "18000",
        "invalid", "51000", "16000", "12500", "49000",
        "17000", "53000", "11000", "16000", "50000"
    ],
    "Order_Date": [
        "2026-01-05", "05/01/2026", "2026-01-07", "10-01-2026",
        "2026/01/12", "15/01/2026", "2026-01-18", "20-01-2026",
        None, "2026/01/22", "25/01/2026", "2026-01-27",
        "30-01-2026", "2026/02/02", "2026-02-05"
    ],
    "Rating": [
        4.5, 4.0, 5.0, 4.2, 3.8, 4.0, 4.7, None,
        3.5, 4.1, 4.0, 4.8, 3.9, 4.3, 5.0
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Statistical Analysis

print(df.isnull())
print(df.isna())
print(df.isnull().sum())
print(df.isna().sum().sum())

df_dropna = df.dropna()
print(df_dropna)
```

```python
#Identify and Handle Missing Values
clean_df = df.copy()

clean_df["City"] = clean_df["City"].fillna("Unknown")
clean_df["Quantity"] = clean_df["Quantity"].fillna(
    clean_df["Quantity"].median()
)
clean_df["Rating"] = clean_df["Rating"].fillna(
    clean_df["Rating"].median()
)
```

```python
price_before = pd.to_numeric(df["Price"], errors="coerce")

print("Total Quantity:", df["Quantity"].sum())
print("Average Quantity:", df["Quantity"].mean())
print("Median Quantity:", df["Quantity"].median())

print("Average Price:", price_before.mean())
print("Median Price:", price_before.median())

print(df[["Quantity", "Rating"]].describe())
```

```python
#Data Formatting and Standardization
clean_df["Customer_Name"] = clean_df["Customer_Name"].str.strip()
clean_df["City"] = clean_df["City"].str.strip()

clean_df["Customer_Name"] = clean_df["Customer_Name"].str.title()
clean_df["City"] = clean_df["City"].str.title()
clean_df["Product"] = clean_df["Product"].str.title()

clean_df["Price"] = pd.to_numeric(
    clean_df["Price"], errors="coerce"
)

clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(clean_df)
```

```python
#Data Type Conversion
print(df.dtypes)
df.info()

clean_df["Order_ID"] = clean_df["Order_ID"].astype("int64")

clean_df["Quantity"] = clean_df["Quantity"].round().astype("Int64")

clean_df["Price"] = pd.to_numeric(
    clean_df["Price"], errors="coerce"
)

clean_df["Rating"] = pd.to_numeric(
    clean_df["Rating"], errors="coerce"
)

print(clean_df.dtypes)
clean_df.info()
```

```python
#Identify Outliers Using IQR
print(clean_df[["Quantity", "Price"]].describe())

Q1 = clean_df["Quantity"].quantile(0.25)
Q3 = clean_df["Quantity"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = clean_df[
    (clean_df["Quantity"] < lower_bound) |
    (clean_df["Quantity"] > upper_bound)
]

print(outliers)
```

```python
#Handle Outliers
before_outlier_handling = clean_df.copy()

clean_df.loc[
    clean_df["Order_ID"] == 1015, "Quantity"
] = 10

print(before_outlier_handling[
    before_outlier_handling["Order_ID"] == 1015
])

print(clean_df[
    clean_df["Order_ID"] == 1015
])
```

```python
#Verify and Compare Results
clean_df = clean_df.drop_duplicates(subset=["Order_ID"])

print("Original Row Count:", len(df))
print("Cleaned Row Count:", len(clean_df))

print("Original Missing Values:")
print(df.isna().sum())

print("Cleaned Missing Values:")
print(clean_df.isna().sum())

print("Original Quantity Statistics:")
print(df["Quantity"].describe())

print("Cleaned Quantity Statistics:")
print(clean_df["Quantity"].describe())

print("Original Price Statistics:")
print(pd.to_numeric(df["Price"], errors="coerce").describe())

print("Cleaned Price Statistics:")
print(clean_df["Price"].describe())

display(df)
display(clean_df)
```

```python
#Compare Statistics Before and After Cleaning
comparison = pd.DataFrame({
    "Measure": [
        "Total Quantity",
        "Mean Quantity",
        "Median Quantity",
        "Mean Price",
        "Median Price"
    ],
    "Before Cleaning": [
        df["Quantity"].sum(),
        df["Quantity"].mean(),
        df["Quantity"].median(),
        pd.to_numeric(df["Price"], errors="coerce").mean(),
        pd.to_numeric(df["Price"], errors="coerce").median()
    ],
    "After Cleaning": [
        clean_df["Quantity"].sum(),
        clean_df["Quantity"].mean(),
        clean_df["Quantity"].median(),
        clean_df["Price"].mean(),
        clean_df["Price"].median()
    ]
})

print(comparison.round(2))
```

```python
#Sales Analysis
clean_df["Sales"] = clean_df["Quantity"] * clean_df["Price"]

sales_summary = clean_df.groupby(
    ["City", "Product"], dropna=False
)["Sales"].sum().reset_index()

print(sales_summary)
```
