```python
9 - Clean duplicate records
```

```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
10 - Handle missing dates
```

```python
df_clean["Date"] = pd.to_datetime(
    df_clean["Date"], errors="coerce"
)
df_clean = df_clean.dropna(subset=["Date"]).copy()

print("Rows after handling missing dates:", len(df_clean))
```

```python
11 — Handle missing categorical values
```

```python
# Replace missing text values with explicit labels
df_clean["City"] = df_clean["City"].fillna("Unknown")
df_clean["Product"] = df_clean["Product"].fillna("Unknown Product")

print("Missing categorical values handled.")
```

```python
12 — Handle missing numerical values
```

```python
numeric_cols = ["Units_Sold", "Unit_Price", "Discount_Pct"]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("Missing numerical values handled.")
```

```python
13 — Verify the cleaning
```

```python
print("Missing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nCleaned dataset shape:", df_clean.shape)
```

```python
Task 2 — Descriptive Statistics
Cell 14 — Generate summary statistics
```

```python
df_clean.describe()
```

```python
15 — Calculate revenue
```

```python
df_clean["Revenue"] = (
    df_clean["Units_Sold"]
    * df_clean["Unit_Price"]
    * (1 - df_clean["Discount_Pct"] / 100)
)

df_clean[["Order_ID", "Units_Sold", "Unit_Price",
          "Discount_Pct", "Revenue"]].head()
```

```python
16 — Calculate mean and total revenue
```

```python
mean_revenue = df_clean["Revenue"].mean()

total_revenue = df_clean["Revenue"].sum()

print(f"Mean revenue: ₹{mean_revenue:,.2f}")
print(f"Total revenue: ₹{total_revenue:,.2f}")
```

```python
Task 3 — Product Analysis
Cell 17 — Count product occurrences
```

```python
product_counts = df_clean["Product"].value_counts()

display(product_counts.to_frame(name="Occurrence_Count"))
```

```python
18 — Calculate total units sold by product
```

```python
units_by_product = (
    df_clean.groupby("Product")["Units_Sold"]
    .sum()
    .sort_values(ascending=False)
)

display(units_by_product.to_frame(name="Total_Units_Sold"))
```

```python
19 — Find the best-selling product
```

```python
best_product = units_by_product.idxmax()
best_units = units_by_product.max()

print("Best-selling product:", best_product)
print("Total units sold:", best_units)
```

```python
Task 4 — Regional Revenue Analysis
Cell 20 — Aggregate regional sales
```

```python
regional_analysis = (
    df_clean.groupby("Region")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Average_Order_Revenue=("Revenue", "mean"),
        Order_Count=("Order_ID", "nunique"),
        Total_Units_Sold=("Units_Sold", "sum")
    )
    .sort_values("Total_Revenue", ascending=False)
)

display(regional_analysis.round(2))
```

```python
21 — Identify the highest-revenue region
```

```python
best_region = regional_analysis["Total_Revenue"].idxmax()

print("Highest-revenue region:", best_region)
print(
    "Revenue:",
    f"₹{regional_analysis.loc[best_region, 'Total_Revenue']:,.2f}"
)
```

```python
22 — Visualize regional revenue
```

```python
# Create a bar chart of regional revenue
regional_analysis["Total_Revenue"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
Task 5 — Correlation Analysis
Cell 23 — Generate the correlation matrix
```

```python
sales_numeric = df_clean[
    ["Units_Sold", "Unit_Price", "Discount_Pct", "Revenue"]
]

correlation_matrix = sales_numeric.corr()

display(correlation_matrix.round(3))
```

```python
24 — Calculate the specific correlation
```

```python
units_revenue_corr = df_clean["Units_Sold"].corr(
    df_clean["Revenue"]
)

print(
    "Correlation between units sold and revenue:",
    round(units_revenue_corr, 3)
)
```

```python
Task 6 — Scatter Plot
Cell 25 — Plot units sold against revenue
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    df_clean["Units_Sold"],
    df_clean["Revenue"],
    alpha=0.7
)

plt.title("Units Sold vs Revenue")
plt.xlabel("Units Sold")
plt.ylabel("Revenue (₹)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
Task 7 — Monthly Revenue Analysis
Cell 26 — Create a month column
```

```python
df_clean["Month"] = df_clean["Date"].dt.to_period("M").astype(str)

df_clean[["Date", "Month", "Revenue"]].head()
```

```python
27 — Calculate monthly revenue
```

```python
monthly_revenue = (
    df_clean.groupby("Month")["Revenue"]
    .sum()
    .sort_index()
)

display(monthly_revenue.to_frame(name="Total_Revenue").round(2))
```

```python
28 — Plot the monthly trend
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Revenue Trend — January to June 2026")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
29 — Find the highest-revenue month
```

```python
best_month = monthly_revenue.idxmax()
best_month_revenue = monthly_revenue.max()

print("Highest-revenue month:", best_month)
print(f"Revenue: ₹{best_month_revenue:,.2f}")
```

```python
Task 8 — Top Five Revenue Orders
Cell 30 — Display the five highest-revenue records
```

```python
top_5_orders = (
    df_clean.nlargest(5, "Revenue")[
        [
            "Order_ID",
            "Date",
            "City",
            "Region",
            "Category",
            "Product",
            "Units_Sold",
            "Unit_Price",
            "Discount_Pct",
            "Revenue"
        ]
    ]
    .reset_index(drop=True)
)

display(top_5_orders.round(2))
```

```python
Cell 31 — Generate a concise summary
```

```python
print("SALES DATA ANALYSIS SUMMARY:")

print("Cleaned records:", len(df_clean))
print(f"Total revenue: ₹{df_clean['Revenue'].sum():,.2f}")
print(f"Mean revenue: ₹{df_clean['Revenue'].mean():,.2f}")
print("Best-selling product:", units_by_product.idxmax())
print("Highest-revenue region:", regional_analysis["Total_Revenue"].idxmax())
print("Highest-revenue month:", monthly_revenue.idxmax())
print("Top revenue order:", df_clean.loc[df_clean["Revenue"].idxmax(), "Order_ID"])
print(
    "Units-revenue correlation:",
    round(df_clean["Units_Sold"].corr(df_clean["Revenue"]), 3)
)
```
